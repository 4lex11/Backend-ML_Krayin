```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True
)

print("Primera compra:", df["Order Date"].min())
print("Última compra:", df["Order Date"].max())
```

```python
print("Registros:", len(df))
print("Clientes únicos:", df["Customer ID"].nunique())
print("Pedidos únicos:", df["Order ID"].nunique())
print("Productos únicos:", df["Product ID"].nunique())
```

```python
clientes = (
    df.groupby("Customer ID")
      .agg(
          compras=("Order ID", "nunique"),
          ventas_totales=("Sales", "sum"),
          primera_compra=("Order Date", "min"),
          ultima_compra=("Order Date", "max")
      )
      .reset_index()
)

clientes.head()
```

```python
fecha_referencia = df["Order Date"].max()

rfm = (
    df.groupby("Customer ID")
      .agg(
          Recency=("Order Date", lambda x: (fecha_referencia - x.max()).days),
          Frequency=("Order ID", "nunique"),
          Monetary=("Sales", "sum")
      )
      .reset_index()
)

rfm.head()

rfm.describe()
```

```python
rfm.info()

print("Clientes en RFM:", len(rfm))
```

```python
fecha_corte = pd.Timestamp("2018-06-30")

print("Fecha de corte:", fecha_corte)

datos_historial = df[df["Order Date"] <= fecha_corte].copy()

datos_futuro = df[df["Order Date"] > fecha_corte].copy()

print("Registros históricos:", len(datos_historial))
print("Registros futuros:", len(datos_futuro))
```

```python
rfm_modelo = (
    datos_historial.groupby("Customer ID")
    .agg(
        Recency=("Order Date",
                 lambda x: (fecha_corte - x.max()).days),
        Frequency=("Order ID", "nunique"),
        Monetary=("Sales", "sum")
    )
    .reset_index()
)

rfm_modelo.head()
```

```python
clientes_futuro = (
    datos_futuro.groupby("Customer ID")["Order ID"]
    .nunique()
    .reset_index(name="pedidos_futuros")
)

clientes_futuro.head()
```

```python
dataset_ml = rfm_modelo.merge(
    clientes_futuro,
    on="Customer ID",
    how="left"
)

dataset_ml["pedidos_futuros"] = dataset_ml["pedidos_futuros"].fillna(0)

dataset_ml["Churn"] = (
    dataset_ml["pedidos_futuros"] == 0
).astype(int)
```

```python
print(dataset_ml["Churn"].value_counts())

print(dataset_ml["Churn"].value_counts(normalize=True))

```

```python
dataset_ml["Churn"] = (
    dataset_ml["pedidos_futuros"] == 0
).astype(int)

X = dataset_ml[[
    "Recency",
    "Frequency",
    "Monetary"
]]

y = dataset_ml["Churn"]

print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)

print("\nVariables de entrada:")
print(X.head())

print("\nVariable objetivo:")
print(y.head())
```
