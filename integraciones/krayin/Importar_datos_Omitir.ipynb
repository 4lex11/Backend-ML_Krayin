```python
# Ejecutar desde la raíz del proyecto o desde integraciones/krayin.
from pathlib import Path

PROJECT_ROOT = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / "ML" / "Datasets" / "train.csv").is_file()),
    None,
)
if PROJECT_ROOT is None:
    raise RuntimeError("Abra el notebook dentro de la carpeta del proyecto.")
ML_DIR = PROJECT_ROOT / "ML"
INTEGRATION_DIR = PROJECT_ROOT / "integraciones" / "krayin"

import json
import pandas as pd

# 1. Cargar el dataset original
df = pd.read_csv(ML_DIR / "predicciones_crm_churn.csv")

# 2. Nombre e ID
if "name" not in df.columns:
    df["name"] = df["customer_id"].astype(str)

# 3. Correos electrónicos estructurados como JSON
df["emails"] = df["customer_id"].apply(
    lambda x: json.dumps([{"value": f"cliente_{x}@ejemplo.com", "label": "work"}])
)

# 4. Números telefónicos ÚNICOS estructurados como JSON (resuelve el error de duplicados)
df["contact_numbers"] = [
    json.dumps([{"value": f"900{idx:06d}", "label": "work"}])
    for idx in range(1, len(df) + 1)
]

# 5. Asegurar que 'monto_total_gastado' sea de tipo decimal puro sin valores nulos
df["monto_total_gastado"] = (
    pd.to_numeric(df["monto_total_gastado"], errors="coerce")
    .fillna(0.0)
    .round(2)
)

# 6. ID del usuario asignado
df["user_id"] = 1

# 7. Guardar el archivo listo
df.to_csv(INTEGRATION_DIR / "predicciones_crm_churn_perfecto.csv", index=False)
print("¡Archivo 'predicciones_crm_churn_perfecto.csv' generado con éxito!")
```
